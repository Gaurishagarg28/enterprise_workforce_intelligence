# 03 — Feature Engineering
Prepare the validated HR dataset for attrition modeling while excluding identifiers and constant columns.

```python
from pathlib import Path
import pandas as pd
df = pd.read_csv(Path('../data/raw/employee_attrition.csv'))
target = df['Attrition'].map({'No':0,'Yes':1})
drop_cols = ['Attrition','EmployeeNumber','EmployeeCount','Over18','StandardHours']
X = df.drop(columns=drop_cols)
X = pd.get_dummies(X, drop_first=True)
print('Features:', X.shape)
print('Target positive rate:', target.mean())
```

```python
print(X.dtypes.value_counts())
print('Any missing:', X.isnull().sum().sum())
```
