# 06 — Agentic Workforce Intelligence Demo
Demonstrate the implemented specialist-agent workflow: attrition risk → skill gap → recommendation, with optional transformer and RAG layers.

```python
import pandas as pd
from pathlib import Path
from src.agent_graph import build_workforce_graph
from src.train_model import MODEL_FEATURES, train_attrition_model
df=pd.read_csv(Path('../data/raw/employee_attrition.csv'))
model,_=train_attrition_model(df)
employee=df.iloc[[0]]
graph=build_workforce_graph()
result=graph.invoke({'model':model,'employee':employee,'features':MODEL_FEATURES,'role':str(employee['JobRole'].iloc[0]),'current_skills':['Communication']})['intelligence']
result
```

## Interpretation
The agents do not make autonomous employment decisions. The model supplies a risk signal, the skill agent calculates a controlled capability gap, and the recommendation agent applies transparent rules. Human review remains mandatory.
