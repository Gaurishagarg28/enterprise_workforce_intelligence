# 01 — Data Exploration
Explore the validated IBM HR Analytics dataset and understand workforce attrition patterns.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
DATA_PATH = Path('../data/raw/employee_attrition.csv')
df = pd.read_csv(DATA_PATH)
print(df.shape)
df.head()
```

```python
print(df['Attrition'].value_counts())
print((df['Attrition'].value_counts(normalize=True)*100).round(2))
print(df.groupby('Department')['Attrition'].apply(lambda s: (s == 'Yes').mean()*100).round(2))
```
