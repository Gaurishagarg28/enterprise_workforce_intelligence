# 07 — O*NET Skill & Role Intelligence
Use the supplied O*NET occupation, essential-skill, and software-skill datasets to build a defensible role capability layer.

This notebook does not invent employee skills. It creates a benchmark of what a role requires and exposes gaps between available employee attributes and validated role requirements.

```python
from pathlib import Path
import pandas as pd
ROOT = Path('..')
raw = ROOT / 'data' / 'raw'
occupations = pd.read_csv(raw / 'occupation_data.csv')
software = pd.read_csv(raw / 'software_skills.csv')
essential = pd.read_csv(raw / 'essential_skills.csv')
print('Occupations:', occupations.shape)
print('Software skills:', software.shape)
print('Essential skills:', essential.shape)
```

```python
role = 'Data Scientists'
role_rows = occupations[occupations['Title'].str.contains(role, case=False, regex=False)]
role_rows[['O*NET-SOC Code','Title','Description']].head()
```

```python
role_software = software[software['Title'].str.contains(role, case=False, regex=False)].copy()
role_software[['Workplace Example','Element Name','Hot Technology','In Demand']].head(20)
```

```python
role_essential = essential[essential['Title'].str.contains(role, case=False, regex=False)].copy()
role_essential[role_essential['Scale ID'].eq('IM')][['Element Name','Data Value']].sort_values('Data Value', ascending=False).head(20)
```

## How this feeds the product
O*NET becomes the external benchmark. The employee dataset supplies workforce/attrition signals; O*NET supplies role requirements. The agentic layer can then recommend development actions against explicit evidence instead of generating arbitrary skills.
